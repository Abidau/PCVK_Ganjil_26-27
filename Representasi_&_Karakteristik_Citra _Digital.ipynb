````python
img = cv2.imread('kartu.png')

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

total_pixel = gray.size

dark_pixel = np.sum(gray < 100)

persentase = (dark_pixel / total_pixel) * 100

print("ukuran citra grayscale: ", gray.shape)
print("jumlah seluruh pixel: ", total_pixel)
print("jumlah pixel dengan nilai < 100: ", dark_pixel)
print("persentase pixel dengan nilai < 100: ", persentase, "%")

plt.figure(figsize=(10, 5))
plt.imshow(gray, cmap='gray')
plt.title("Citra Grayscale")
plt.axis("off")
plt.show()
````

````python
def label(foto, nama):

    img = cv2.imread(foto)
    tinggi, lebar = img.shape[:2]

    bar = int(tinggi * 0.1)
    img[tinggi-bar:tinggi, :] = (0, 0, 0)

    ukuran = tinggi * 0.002

    cv2.putText(
        img,
        nama,
        (int(lebar * 0.03), int(tinggi * 0.96)),
        cv2.FONT_HERSHEY_SIMPLEX,
        ukuran,
        (255, 255, 255),
        2
    )

    return img


hasil1 = label("foto1.jpeg", "Naufal Abid Aurizky")
hasil2 = label("foto2.jpeg", "Naufal Abid Aurizky")

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(hasil1, cv2.COLOR_BGR2RGB))
plt.title("Foto 1")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(hasil2, cv2.COLOR_BGR2RGB))
plt.title("Foto 2")
plt.axis("off")

plt.show()
````

````python
img = cv2.imread("produk.jpeg")

tinggi, lebar = img.shape[:2]
sisi = max(tinggi, lebar)

canvas = np.full(
    (sisi, sisi, 3),
    255,
    dtype=np.uint8
)

x = (sisi - lebar) // 2
y = (sisi - tinggi) // 2

canvas[y:y+tinggi, x:x+lebar] = img

hasil = cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(7, 7))
plt.imshow(hasil)
plt.axis("off")
plt.show()
````
