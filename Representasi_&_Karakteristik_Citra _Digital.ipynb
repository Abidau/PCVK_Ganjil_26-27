Membuat bentuk Geometri 2D dari OpenCV

````python
black_img = np.zeros((530, 530, 3), dtype=np.int16)
plt.imshow(black_img)
````

Menambahkan sebuah persegi panjang sesuai koordinat pt1 dan pt2

````python
cv.rectangle(black_img,pt1=(290,0),pt2=(520,130),color=(0,0,255),thickness=10)
plt.imshow(black_img)
````

Membuat bentuk Persegi sesuai koordinat pt1 dan pt2

````python
cv.rectangle(black_img,pt1=(250,250),pt2=(350,350),color=(255,0,0),thickness=15)
plt.imshow(black_img)
````

Menambahkan bentuk lingkaran sesuai radius yang tertulis pada praktikum

````python
cv.circle(black_img,center=(150,150),radius=60,color=(0,255,0),thickness=8)
plt.imshow(black_img)
````

Menambahkan lingkaran berwarna dan garis sesuai koordinat pt1 dan pt2

````python
cv.circle(black_img,center=(400,400),radius=50,color=(0,255,0),thickness=-1)
cv.line(black_img,pt1=(0,0),pt2=(530,530),color=(255,255,255),thickness=5)
plt.imshow(black_img)
````

Menambahkan text dengan font yang telah tertulis dengan ukuran yang sudah ditentukan pada praktikum

````python
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img,text='PCVK is FUN!', org=(10,500), fontFace=font, fontScale=2,
           color=(255,255,0), thickness=3, lineType=cv.LINE_AA)
plt.imshow(black_img)
````

Membuat sebuah black image kembali dilakukan dengan tipe data int32

````python
black_img = np.zeros((530, 530, 3), dtype=np.int32)
plt.imshow(black_img)
````

Melakukan kode program untuk inisialisasi NumPy array dengan tipe data int32

````python
vertices = np.array([[150, 300], [250, 250], [450, 350], [250, 450]], dtype=np.int32)
````

Array tersebut di reshape sebagai berikut

````python
pts = vertices.reshape((-1,1,2))
pts
````

Lalu penambahan polyline pada black image kedua yang telah dibuat

````python
cv.polylines(black_img, [pts], isClosed=True, color=(0,250,0), thickness=5)
plt.imshow(black_img)
````
